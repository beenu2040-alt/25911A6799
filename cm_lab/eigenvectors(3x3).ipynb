```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("enter the order of the matrix: "))
print("enter the elements row wise")
A=[]
for i in range(n):
    row=list(map(complex,input(">").split()))
    A.append(row)
A=np.array(A,dtype=complex)
eigenvalues,eigenvectors=np.linalg.eig(A)
eigenvalues=np.real_if_close(eigenvalues) 
eigenvectors=np.real_if_close(eigenvectors)
eigenvalues=np.round(eigenvalues,4)
eigenvectors=np.round(eigenvectors,4)
print(eigenvectors)
```

```python
ax=plt.axes(projection='3d')
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("eigrnvectors of a 3x3 matrix")
colors=['red','blue','black']
for i in range(n):
    ax.quiver(0,0,0,eigenvectors[0,i],eigenvectors[1,i],eigenvectors[2,i],color=colors[i],label=f'eigenvectors{i+1}',length=1,normalize=False)
plt.legend()
plt.show()
```
