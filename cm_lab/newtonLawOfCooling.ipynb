```python
import numpy as np
import matplotlib.pyplot as plt


Ts = float(input('enter surrounding temperature in C: '))
t1=float(input('enter first time t1 in minutes: '))
T1=float(input('enter temperature at first time(t1): '))
t2=float(input('enter first second t2 in minutes: '))
T2=float(input('enter temperature at second time(t2): '))
k=np.log((T1-Ts)/(T2-Ts))/(t2-t1)
C=(T1-Ts)*np.exp(k*t1)
print('\nselect an option')
print('1- find temperature at a given time')
print('2-find timetaken to reach a given temperature')
choice=int(input('enter your choice(1 or 2): '))
if choice==1:
    t=float(input('enter time in minutes'))
    T=Ts +C*np.exp(-k*t)
    print('Temperature at time ',t,'=',round(T,2))
elif choice ==2:
    T=float(input('enter target temperature: '))
    t=-np.log((T-Ts)/C)/k
    print('time required to reach ',T,' temperature is ',round(t,2))
else :
    print('invalid choice')
plt.figure(figsize=(9,5))
time_values=np.linspace(0,t*1.2,300)
temp_values=Ts+C*np.exp(-k*time_values)
plt.xlabel('time')
plt.ylabel('temp')
plt.title('newton law of cooling')
plt.grid()
plt.plot(time_values,temp_values,label='Cooling curve')
plt.scatter([t1,t2],[T1,T2],color='red',label='observed points')
plt.scatter(t,T,color='green',s=70,label='calculated point')
plt.axhline(Ts,linestyle='--',label='surrounding temp')
plt.legend()
plt.show()
```
