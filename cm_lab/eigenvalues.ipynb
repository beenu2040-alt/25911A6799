```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("enter the order of the matrix: "))
print("enter the elements row wise")
A=[]
for i in range(n):
    row=list(map(complex,input(">").split()))
    A.append(row)
A=np.array(A,dtype=complex)
eigenvalues=np.linalg.eigvals(A)
eigenvalues=np.real_if_close(eigenvalues) 
eigenvalues=np.round(eigenvalues,4)
print(eigenvalues)
plt.axhline(0)
plt.axvline(0)
plt.xlabel("Real part")
plt.ylabel("imaginary part")
plt.grid("true")
plt.title("eigenvalues of the matrix")
plt.scatter(eigenvalues.real,eigenvalues.imag,s=80)
plt.show()
```
