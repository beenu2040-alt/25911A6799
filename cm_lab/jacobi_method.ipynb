```python
import numpy as np
import matplotlib.pyplot as plt

# Read number of equations
n = int(input("Enter the number of equations: "))

# Read coefficient matrix
print("Enter the coefficient matrix row by row:")
A = np.array([list(map(float, input().split())) for i in range(n)])

# Read constants
print("Enter the constants:")
B = np.array(list(map(float, input().split())))

# Check diagonal dominance
flag = True

for i in range(n):
    s = 0

    for j in range(n):
        if i != j:
            s = s + abs(A[i][j])

    if abs(A[i][i]) <= s:
        flag = False
        break

if flag:
    print("\nThe matrix is diagonally dominant.")
    print("Jacobi Method can be applied.")

    # Initial guess
    X = np.zeros(n)

    # Tolerance
    tolerance = float(input("Enter the tolerance: "))

    # Jacobi iteration
    while True:
        X_new = np.zeros(n)

        for i in range(n):
            s = 0

            for j in range(n):
                if i != j:
                    s = s + A[i][j] * X[j]

            X_new[i] = (B[i] - s) / A[i][i]

        # Check convergence
        if np.max(np.abs(X_new - X)) < tolerance:
            break

        X = X_new.copy()

    # Display solution
    print("\nSolution:")
    for i in range(n):
        print(f"X{i + 1} = {X_new[i]:.4f}")

    # 3D visualization for 3x3 system
    if n == 3:

        fig = plt.figure()
        ax = fig.add_subplot(111, projection='3d')

        ax.set_xlabel("X")
        ax.set_ylabel("Y")
        ax.set_zlabel("Z")

        x = np.linspace(X_new[0] - 2, X_new[0] + 2, 20)
        y = np.linspace(X_new[1] - 2, X_new[1] + 2, 20)

        x1, y1 = np.meshgrid(x, y)

        plt.title("Jacobi Method - Solution")

        colors = ('red', 'blue', 'green')

        for i in range(3):
            a = A[i][0]
            b = A[i][1]
            c = A[i][2]
            d = B[i]

            if c != 0:
                z = (d - a * x1 - b * y1) / c

                ax.plot_surface(
                    x1,
                    y1,
                    z,
                    alpha=0.5,
                    color=colors[i]
                )

        # Mark solution point
        ax.scatter(
            X_new[0],
            X_new[1],
            X_new[2],
            color='black',
            s=80
        )

        ax.text(
            X_new[0],
            X_new[1],
            X_new[2],
            "Solution"
        )

        plt.show()

    else:
        print("\nVisualization is available only for a 3x3 system.")

else:
    print("\nThe matrix is NOT diagonally dominant.")
    print("Jacobi Method may not converge.")
```
