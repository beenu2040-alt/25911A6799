```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
x,y = sp.symbols('x y')
M=sp.sympify(input('enter M(x,y)'))
N=sp.sympify(input('enter N(x,y)'))
My=sp.diff(M,y)
Nx=sp.diff(N,x)
if sp.simplify(My-Nx)==0:
    F=sp.integrate(M,x)
    g_prime=sp.simplify(N-sp.diff(F,y))
    g=sp.integrate(g_prime,y)
    F=sp.simplify(F+g)
    print('\ngeneral solution')
    print(F,'=C')
    f_func=sp.lambdify((x,y),F,'numpy')
    x_values=np.linspace(-5,5,300)
    y_values=np.linspace(-5,5,300)
    x,y=np.meshgrid(x_values,y_values)
    z=f_func(x,y)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('solution curve')
    plt.contour(x,y,z,levels=15)
    plt.grid()
    plt.show()


else:
    print("\nequation is not exact")
    
```
