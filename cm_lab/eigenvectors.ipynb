```python
import numpy as np
import matplotlib.pyplot as plt
n=int(input("enter the order of the matrix: "))
print("enter the elements row wise")
A=[]
for i in range(n):
    row=list(map(complex,input(">").split()))
    A.append(row)
A=np.array(A,dtype=complex)
eigenvalues,eigenvectors=np.linalg.eig(A)
eigenvalues=np.real_if_close(eigenvalues) 
eigenvectors=np.real_if_close(eigenvectors)
eigenvalues=np.round(eigenvalues,4)
eigenvectors=np.round(eigenvectors,4)
print(eigenvalues)
print(eigenvectors)
```

```python
plt.axhline(0)
plt.axvline(0)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.title("Eigen vectors of a 2x2 matrix")
colors=['red','black']
for i in range(n):
    plt.quiver(0,0,eigenvectors[0,i],eigenvectors[1,i],scale=3,color=colors[i],label= f'Eigenvector{i+1}')
plt.legend()
```
